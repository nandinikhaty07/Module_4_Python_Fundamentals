## 1. Boolean logic is a decision basis (10 min)

A condition is an expression that evaluates to `True` or `False`. We build them using comparisons and logical operators:

- `and`: both conditions must be true
- `or`: at least one condition must be true
- `not`: reverses a Boolean value

```python
temperature_c = 85
pressure_bar = 4.8

print(temperature_c > 80)
print(temperature_c > 80 and pressure_bar < 5.0)
print(temperature_c > 100 or pressure_bar > 6.0)
print(not (pressure_bar > 6.0))
```

## 2. The `if` statement (15 min)

An `if` block runs only when its condition is true. Python uses **indentation** (usually four spaces) to show which lines belong to the block. Indentation is part of Python syntax, not just formatting.

```python
tank_level_percent = 80

if tank_level_percent > 85:
    print('High tank level: check outlet flow.')

    print('Monitoring continues.')
```

## 3. `if` / `else` / `elif` (15 min)

Use `else` for the alternative action. Use `elif` when there are several mutually exclusive ranges. Put more restrictive conditions first when ranges overlap.

```python
ph = 4

if ph < 6.0:
    status = 'Acidic: consider adding base.'
elif ph <= 8.0:
    status = 'Target range.'
else:
    status = 'Alkaline: consider adding acid.'

print(f'pH {ph}: {status}')
```

## 4. Decision logic for a pump (12 min)

Here is a simplified interlock rule: a pump may start only when the tank level is above 20% **and** the emergency stop is not active. This is a teaching example, not a replacement for safety-rated control systems.

```python
tank_level_percent = 35
emergency_stop_active = False

if tank_level_percent > 20 and not emergency_stop_active:
    print('Pump start permitted.')
else:
    print('Pump start blocked: inspect conditions.')
```
